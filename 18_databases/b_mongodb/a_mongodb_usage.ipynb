```python
# Create a collection
collection = db['mycollection']
collection
```

```python
# Insert a Document

document = {"name": "Alice", "age": 25}
collection.insert_one(document)
```

```python
# Insert Multiple Documents

documents = [
        {"name": "Bob", "age": 30}, 
        {"name": "Charlie", "age": 35}
]
collection.insert_many(documents)
```

```python
collection.insert_one({
    # "_id" : ObjectId("679e31a5bac09a8e39696a56"),
    "name" : "Bob",
    "age" : 30
})
```

```python
# Find a Document

result = collection.find_one({"name": "Alice"})
print(result)
```

```python
result = collection.find_one({"name": "Bob"})
print(result)
```

```python
result = collection.find({"name": "Bob"})
print(result)
```

```python
for doc in collection.find({"name": "Bob"}):
    print(doc)

# db.getCollection('mycollection').find({"name": "Bob"})
```

```python
# Find All Documents
for doc in collection.find(): # no condition means all 
    print(doc)
```

```python
# Count Documents
count = collection.count_documents({})
print(count)
```

```python
# Update a Document
collection.update_one(
        {"name": "Alice"}, 
        {"$set": {"age": 26}}
        )

# update table set ag=26 where name= 'Alice' 
```

```python

result = collection.find_one({"name": "Alice"})
print(result)
```

```python
collection.update_one(
        {"name": "applePie"}, 
        {"$set": {"age": 26}}
        )
```

```python
# Update Multiple Documents

collection.update_many(
        {"age": {"$gt": 30}},            # action
        {"$set": {"status": "senior"}}   # condition
        )

for doc in collection.find(): # no condition means all 
    print(doc)
```

```python
# Delete a Document
collection.delete_one({"name": "Alice"})
```

```python
# Delete Multiple Documents
collection.delete_many({"age": {"$lt": 30}})
```

```python
# Find with Conditions
for doc in collection.find({"age": {"$gte": 30}}):
    print(doc)
```

```python
# Sort Documents

for doc in collection.find().sort("age"):
    print(doc)
```

```python
# Limit the Number of Results
for doc in collection.find().limit(2):
    print(doc)
```

```python
# Projection (Select Fields)
for doc in collection.find():
    print(doc)
```

```python
for doc in collection.find({}, {"name": 1}):
    print(doc)
```

```python
# Use Regular Expressions
for doc in collection.find({"name": {"$regex": "^B"}}):
    print(doc)
```

```python
# Using Aggregation

pipeline = [
    {"$group": {"_id": "$age", "count": {"$sum": 1}}}
]
result = collection.aggregate(pipeline)
for doc in result:
    print(doc)
```

```python
# Indexing

collection.create_index([("name", 1)])
```

```python
# Check Indexes
indexes = collection.index_information()
print(indexes)
```

```python
# Using $or Operator
query = {"$or": [{"name": "Bob"}, {"age": 35}]}
for doc in collection.find(query):
    print(doc)
```

```python
# Using $and Operator
query = {"$and": [{"age": {"$gt": 25}}, {"age": {"$lt": 35}}]}
for doc in collection.find(query):
    print(doc)
```

```python
# Bulk Write Operations

operations = [
    pymongo.InsertOne({"name": "David"}),
    pymongo.UpdateOne({"name": "Bob"}, {"$set": {"age": 31}}),
    pymongo.DeleteOne({"name": "Charlie"})
]
collection.bulk_write(operations)
```

```python
# Using Transactions (MongoDB 4.0+)

with client.start_session() as session:
    with session.start_transaction():
        collection.insert_one({"name": "Eve"}, session=session)
        collection.update_one({"name": "Bob"}, {"$set": {"age": 32}}, session=session)
```

```python
# Change Streams
with collection.watch() as stream:
    for change in stream:
        print(change)
```

```python
# Geospatial Queries
collection.create_index([("location", "2dsphere")])
results = collection.find({
    "location": {
        "$near": {
            "$geometry": {
                "type": "Point",
                "coordinates": [-73.97, 40.77]
            },
            "$maxDistance": 1000
        }
    }
})
```

```python
# Text Search
collection.create_index([("name", "text")])
for doc in collection.find({"$text": {"$search": "Bob"}}):
    print(doc)
```

```python
# Using $lookup for Joins
pipeline = [
    {
        "$lookup": {
            "from": "othercollection",
            "localField": "fieldInThisCollection",
            "foreignField": "fieldInOtherCollection",
            "as": "joined_docs"
        }
    }
]
result = collection.aggregate(pipeline)
```

```python
# Using $group and $sort
pipeline = [
    {"$group": {"_id": "$age", "total": {"$sum": 1}}},
    {"$sort": {"total": -1}}
]
for doc in collection.aggregate(pipeline):
    print(doc)
```

```python
# Using $unwind

pipeline = [
    {"$unwind": "$arrayField"},
    {"$group": {"_id": "$arrayField", "count": {"$sum": 1}}}
]
for doc in  collection.aggregate(pipeline):
    print(doc)
```

```python
# Handling Exceptions
try:
    collection.insert_one({"name": "Frank"})
except Exception as e:
    print("An error occurred:", e)
```

```python
# Use of BSON Types
from bson import ObjectId

doc = {"_id": ObjectId(), "name": "Grace"}
collection.insert_one(doc)
```

```python
# Upsert Operation - create if not present, update if present
collection.update_one({"name": "Hannah"}, {"$set": {"age": 28}}, upsert=True)
```

```python
# Using Pagination

page = 1
page_size = 3
for doc in collection.find().skip((page - 1) * page_size).limit(page_size):
    print(doc)
```

```python
# Export to JSON

import json

documents = list(collection.find())
```

```python
try:
    with open('output.json', 'w') as json_file:
        json.dump(documents, json_file)
except TypeError as ex:
    print(ex)
```

```python
import json

class MongoJsonEncoder(json.JSONEncoder):
    def default(self, obj):
        if isinstance(obj, ObjectId):
            return str(obj)  # Convert ObjectId to string
        return super(MongoJsonEncoder, self).default(obj)
    

with open('output.json', 'w') as json_file:
    json.dump(documents, json_file, cls=MongoJsonEncoder)
```
